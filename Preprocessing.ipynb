## Lowercasing

```python
msg = "I LOVE KANDY and I LOVE NLP"

# convert to lowercase
lower_msg = msg.lower()
print(lower_msg)

# count thr number of different words (split on spaces) before and after lowercasing
print(f"Number of words before lowercasing: {len(msg.split())}")
print(f"Number of words after lowercasing: {len(msg.split())}")
```

## Remove Punctuation

```python
sms = "Woww!! Is this really free??? Call now."

# remove the punctuation with a regular expression
print(re.sub(r'[^\w\s]', '',sms))

# with spaCy, list the token that are not punctuation (is_punct) #spaCy
print([t.text for t in nlp(sms) if not t.is_punct])
```

## Removing Numbers

```python
sms = "Win 1000 rupees! Call 0771234567 before 31 Dec"

# remove all numbers and fix the extra spaces
print(re.sub(r'\d+', '', sms))
print(re.sub(r'\d+', '', sms).split())
print(" ".join(re.sub(r'\d+', '', sms).split()))

#insted of removing them replae every number
```

## URLs, emails, HTML tags, @handles and #hashtags

```python
post = "Check https://uni.lk/nlp and mail admin@uni.lk #NLP @anna <b>today</b>"

# remove URL
post = re.sub(r'https?://\S+|www\.\S+', '',post)
print(post)

# remove Email
post = re.sub(r'\S+@\S+', '', post)

# remove tags
post = re.sub(r'<.*?>', '',post)

# remove @handles
post = re.sub(r'@\w+', '',post)

#NLP -> NLP
post = re.sub(r'#(\w+)', r'\1',post)
```
